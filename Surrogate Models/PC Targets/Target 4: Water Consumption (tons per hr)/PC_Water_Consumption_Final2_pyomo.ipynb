```python
# water consumption = cooling-tower evaporation + drift
df['Water Consumption (tons/hr)'] = df['Evaporation (tons/hr)'] + df['Drift Loss (tons/hr)']

# high-ambient rows zero out evaporation (infeasible tower, not real). dropping
df = df[df['Evaporation (tons/hr)'] > 0].reset_index(drop=True)
```

```python
df
```

```python
x = df[['Gross Electrical Output (MWg)',
    'As-Delivered Primary Fuel Cost ($/ton)',
    'Cooling Water Temperature Rise (°F)',
    'Absorber CO2 Removal Efficiency (%)',
    'Amine Scrubber Power Requirement (% MWg)']]

import numpy as np
# fitting directly on the tons/hr values gave poors metrics.
# taking log produced much better metrics; predictions get exp()'d back to tons/hr.
y = np.log(df['Water Consumption (tons/hr)'])
```

```python
from sklearn.model_selection import GridSearchCV, train_test_split, RandomizedSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from xgboost import XGBRegressor
from sklearn.ensemble import GradientBoostingRegressor

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("\n=== PYOMO MODEL ===\n")
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

poly_pipe = Pipeline([
    ('poly',   PolynomialFeatures(include_bias=False)),
    ('scaler', StandardScaler()),
    ('reg',    Ridge())
])
poly_grid = GridSearchCV(
    poly_pipe,
    {'poly__degree': [2, 3, 4],
     'reg__alpha':  [0.001, 0.01, 0.1, 1, 10, 100]},
    cv=5, scoring='r2', n_jobs=-1
)
poly_grid.fit(x_train, y_train)

best_poly = poly_grid.best_estimator_
print("Polynomial best params:", poly_grid.best_params_)
print('Train R2:', best_poly.score(x_train, y_train))
print('Test R2:', best_poly.score(x_test, y_test))
```

```python
#Checking the RMSE and MAPE scores (raw tons/hr, model fit in log)
from sklearn.metrics import mean_squared_error
import numpy as np

print("\n=== RMSE & MAPE (raw tons/hr) ===\n")

models = {
    'Polynomial':        best_poly
}

y_test_raw = np.exp(y_test)                         # y_test is log -> tons/hr

for name, model in models.items():
    y_pred = np.exp(model.predict(x_test))         # predictions -> tons/hr
    rmse   = np.sqrt(mean_squared_error(y_test_raw, y_pred))
    mape   = np.mean(np.abs((y_test_raw - y_pred) / y_test_raw)) * 100
    print(f"{name} | RMSE: {rmse:.4f} | MAPE: {mape:.4f}%")
```

```python
import json, joblib, numpy as np

poly   = best_poly.named_steps['poly']
scaler = best_poly.named_steps['scaler']
reg    = best_poly.named_steps['reg']
feature_names = list(x.columns)

# keep standardization in the equation: unscaling a deg-4 poly over a large input range
# (Gross Output ~2500, ^4 ~ 4e13) loses precision. (m-mean)/scale per term stays stable.
terms = []
for j, p in enumerate(poly.powers_):
    term_powers = {feature_names[i]: int(e) for i, e in enumerate(p) if e > 0}
    terms.append({
        "powers": term_powers,
        "coef":   float(reg.coef_[j]),
        "mean":   float(scaler.mean_[j]),
        "scale":  float(scaler.scale_[j]),
    })

equation = {"intercept": float(reg.intercept_),
            "degree": int(poly.degree),
            "transform": "log",     # predicts log(water) -> optimiser applies exp()
            "scaled": True,         # terms carry mean/scale -> equation_value must standardise
            "terms": terms}

def equation_value(eq, row):
    total = eq["intercept"]
    for t in eq["terms"]:
        m = 1.0
        for feat, e in t["powers"].items():
            m *= row[feat] ** e
        total += t["coef"] * (m - t["mean"]) / t["scale"]
    return total

check = np.array([equation_value(equation, r) for _, r in x.iterrows()])
assert np.allclose(check, best_poly.predict(x), atol=1e-6), "equation does not match model!"
print("equation matches pipeline (log space) ✓")
print("sample tons/hr:", np.round(np.exp(check[:5]), 1))

with open("PC_Water_Consumption_poly_eq.json","w") as f: json.dump(equation, f, indent=2)
print("saved PC_Water_Consumption_poly_eq.json")
joblib.dump(best_poly, "PC_Water_Consumption_poly.pkl")
print("saved PC_Water_Consumption_poly.pkl")
```
