```python
import pandas as pd
df = pd.read_csv("PC_Uncertainty_dataset.csv")
```

```python
df
```

```python
x = df[['Gross Electrical Output (MWg)',
    'As-Delivered Primary Fuel Cost ($/ton)',
    'Cooling Water Temperature Rise (°F)',
    'Absorber CO2 Removal Efficiency (%)',
    'Amine Scrubber Power Requirement (% MWg)']]

y = df['Important Performance & Cost Factors: CO2 Removed (tons/hr)']
```

```python
from sklearn.model_selection import GridSearchCV, train_test_split, RandomizedSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from xgboost import XGBRegressor
from sklearn.ensemble import GradientBoostingRegressor

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("\n=== FINAL MODEL FOR SHAP ===\n")
```

```python
# Gradient Boosting Regressor
gbr_grid = GridSearchCV(
    GradientBoostingRegressor(random_state=42),
    {
        "n_estimators": [100, 300],
        "max_depth": [3, 6, 10],
        "learning_rate": [0.01, 0.1, 0.3],
        "subsample": [0.8, 1.0],
        "max_features": [0.8, 1.0]
    },
    cv=5, scoring="r2", n_jobs=-1
)
gbr_grid.fit(x_train, y_train)

best_gbr = gbr_grid.best_estimator_
print("Gradient Boosting best params:", gbr_grid.best_params_)
print('Train R2:', best_gbr.score(x_train, y_train))
print('Test R2:', best_gbr.score(x_test, y_test))
```

```python
#Checking the RMSE and MAPE scores
from sklearn.metrics import mean_squared_error
import numpy as np

print("\n=== RMSE & MAPE ===\n")

models = {
    'Gradient Boost':    best_gbr
}

for name, model in models.items():
    y_pred = model.predict(x_test)
    rmse   = np.sqrt(mean_squared_error(y_test, y_pred))
    mape   = np.mean(np.abs((y_test - y_pred) / y_test)) * 100
    print(f"{name} | RMSE: {rmse:.4f} | MAPE: {mape:.4f}%")
```

```python
import shap

explainer   = shap.TreeExplainer(best_gbr)
shap_values = explainer.shap_values(x_test)

# 1. bar: mean |SHAP| — overall driver ranking
shap.summary_plot(shap_values, x_test, plot_type="bar")

# 2. beeswarm: direction + magnitude per feature
shap.summary_plot(shap_values, x_test)

# 3. dependence plot for the top driver (swap in whichever ranks #1)
shap.dependence_plot("Absorber CO2 Removal Efficiency (%)", shap_values, x_test)
```

**BEESWARM READ**
- Wide horizontal spread (like Gross Output, −1000 to +750) = big impact — that feature swings the prediction a lot.
- All piled near 0 (like fuel cost) = no impact.

**DEPENDENCE READ**
- e.g. at 65% efficiency, that low efficiency pushed predicted CO2 removed down by ~100 t/hr, versus the average plant (at low gross electrical output, blue).
- The y-axis is efficiency's contribution to the predicted CO2 removed.
- Negative SHAP (y-axis) → feature pulled the prediction down (knocked off).
- Positive SHAP (y-axis) → feature pushed the prediction up (added).
